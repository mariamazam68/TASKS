**Bar Chart vs. Histogram:**

Bar Chart: Use for discrete categories (like "apples" or "bananas"). Bars have clear spaces between them.
Histogram: Use for continuous numerical data (like height or weight) grouped into bins. Bars touch each other to show a continuous range.

**What a Boxplot Shows That a Histogram Misses**
A boxplot shows:

-Median

-Q1 and Q3 (quartiles)

-Spread of the data

-Outliers

-Minimum and maximum values (excluding outliers)

A histogram shows the overall shape and spread of a distribution, but hides exact quartiles and masks individual outliers inside a bin.

**Adding Labels, Title, and Legend in Matplotlib**
xlabel() - labels X-axis

ylabel() - labels Y-axis

title() - adds chart title

legend() - shows what each line/category represents

**Heatmap Use in Exploratory Data Analysis (EDA)**

A heatmap is commonly used to visualize a correlation matrix between numerical variables.The colours make strong and weak relationships easy to identify.

In EDA, heatmaps are especially useful for finding correlations and patterns in data.

**Why is a pie chart usually a bad choice for more than 4–5 categories?**

When there are too many categories, a pie chart becomes crowded and difficult to read.

For example: comparing 12 categories means there will be many small slices, making it harder to tell which slice is larger.

A bar chart is usually better because the values can be compared more easily.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/merged_players.csv")
df
```

**Line chart**

```python
year_avg = df.groupby("year")["avg_fantasy_points"].mean()
plt.figure(figsize=(8, 4))
plt.plot(year_avg.index, year_avg.values, marker="o")
plt.title("Average Fantasy Points Over the Years")
plt.xlabel("Year")
plt.ylabel("Average Fantasy Points")
plt.show()
```

It shows how fantacy points of players changed over years

**Bar chart**

```python
team_avg = df.groupby("team")["avg_fantasy_points"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10, 5))
team_avg.plot(kind="bar")
plt.title("Top 10 Teams by Average Fantasy Points")
plt.xlabel("Team")
plt.ylabel("Average Fantasy Points")
plt.show()
```

This chart compares the fantasy points of teams across top 10 teams

**Histogram**

```python
plt.figure(figsize=(10, 5))
plt.hist(df["avg_fantasy_points"].dropna(), bins=20)
plt.title("Distribution of Average Fantasy Points")
plt.xlabel("Average Fantasy Points")
plt.ylabel("Number of Players")
plt.show()
```

This chart shows how different ranges of fantasy points occur among players.

**Boxplot**

```python
plt.figure(figsize=(10, 5))
plt.boxplot(df["avg_fantasy_points"].dropna())
plt.title("Boxplot of Average Fantasy Points")
plt.ylabel("Average Fantasy Points")
plt.show()
```

This chart shows the median, spread, and unusually high or low fantasy-point values.

**Scatter plot**

```python
plt.figure(figsize=(8, 4))
plt.scatter(df["avg_disposals"], df["avg_fantasy_points"], alpha=0.5)
plt.title("Average Disposals vs Average Fantasy Points")
plt.xlabel("Average Disposals")
plt.ylabel("Average Fantasy Points")
plt.show()
```

This chart shows whether players with more average disposals tend to have higher fantasy points.

**Heatmap**

```python
numeric_cols = [
    "avg_kicks",
    "avg_marks",
    "avg_handballs",
    "avg_disposals",
    "avg_goals",
    "avg_tackles",
    "avg_clearances",
    "avg_fantasy_points"]
correlation = df[numeric_cols].corr()
plt.figure(figsize=(10, 7))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Heatmap of Player Statistics")
plt.show()
```

This heatmap shows the strength and direction of relationships between different player statistics.
