```python
import numpy as np
import pandas as pd
```

**Import Dataset**

```python
players = pd.read_csv("/content/afl_players_info_raw.csv")
stat = pd.read_csv("/content/afl_players_seasonal_stats_raw.csv")
```

**Players info**

```python
print("Players information",players.shape)
print(players.info())
print(players.describe(include="all"),"\n")
```

**Seasonal_stat info**

```python
print("Seasonal_stat information",stat.shape)
print(stat.info())
print(stat.describe(include="all"),"\n")
```

**Checking missing values**

```python
print("Players missing values:\n",players.isnull().sum())
print("\nSeasonal_stat missing values:\n",stat.isnull().sum())
```

**Handling players missing values**

```python
players.drop(columns=["profile_pic"], inplace=True)
players["player_common_names"] = players["player_common_names"].fillna("Unknown")
players["player_teams"] = players["player_teams"].fillna("Unknown")
```

**Handling Seasonal_stat missing values**

```python
numeric_columns = stat.select_dtypes(include="number").columns
exclude_columns = ["year"]
for col in numeric_columns:
    if col not in exclude_columns:
        if stat[col].isnull().sum() > 0:
            median_value = stat[col].median()
            stat[col] = stat[col].fillna(median_value)
            print(f"{col}: missing values filled with median = {median_value}")
```

```python
print("Players missing values:\n",players.isnull().sum())
print("\nSeasonal_stat missing values:\n",stat.isnull().sum())
```

**Checking and handling duplicate values**

```python
print("Duplicate players:", players.duplicated().sum())
print("Duplicate stats:", stat.duplicated().sum())
```

```python
players = players.drop_duplicates()
stat = stat.drop_duplicates()
```

```python
print("Duplicate players:", players.duplicated().sum())
print("Duplicate stats:", stat.duplicated().sum())
```

# Cleaned Players Dataset

```python
players.to_csv("/content/cleaned_players_info.csv",index=False)
print("\nCleaned dataset saved successfully!")
```

# Cleaned Seasonal_stat Dataset

```python
stat.to_csv("/content/cleaned_Seasonal_Stats.csv", index=False)
print("\nSeasonal Stats dataset saved successfully!")
```

**Cleaned datasets info**

```python
print("CLEANED PLAYERS INFO")
print("Nulls:", players.isnull().sum().sum())
print("Duplicates:", players.duplicated().sum())

print("CLEANED SEASONAL STATS")
print("Nulls:", stat.isnull().sum().sum())
print("Duplicates:", stat.duplicated().sum())
```

# Merging cleaned datasets

```python
players = players.rename(columns={"id": "player_id"})

# Merge cleaned datasets
merged_players = pd.merge(players,stat,on="player_id",how="inner")
# Check merged dataset
print("Merged Dataset Shape:", merged_players.shape)
display(merged_players.head())

merged_players.to_csv("/content/merged_players.csv",index=False)
print("Final dataset saved as: /content/merged_players.csv")
```

**Final summary**

```python
# Final Validation Report
validation_report = pd.DataFrame({"Dataset": ["Players","Stat","Merged Dataset"],
                                  "Duplicates Remaining": [players.duplicated().sum(),stat.duplicated().sum(),merged_players.duplicated().sum()],
                                  "Missing Values": [players.isnull().sum().sum(),stat.isnull().sum().sum(),merged_players.isnull().sum().sum()]})
# Display validation report
display(validation_report)
```
