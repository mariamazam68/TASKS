What is the difference between .drop() and .dropna()?

.drop()is used to delete the specific row and column.

.dropna() is used to drop the row and column which contain null values.

How do you change the dtype of a column — and when would you need to?

By using astype we can change the dtype of the column. we need to change the dtype when the column data does not match with the  dtype of the column. For example: age must be in integer values. but if age is in str dtype than it will e changed.

What does .apply() do and how is it different from vectorized operations?

.apply() can run a function on values like a loop. whereas the vectorized operation perform function or operation on entire array at once.

.apply() is used for complex operations or calculations also work slowly whereas the vectorized operations perform simple calculations and is fast.

Difference between .pivot() and .pivot_table()?

.pivot() is used to reshape the entire data but it doesn't handle the duplicates and null values.

.pivot_table is also to reshape the data but it can handle the missing values as well using aggregation operations like mean,median etc.

What does .merge() do and what are the 4 types of joins?

.merge() is used to combine dataframes or datasets.

inner join: It is used to join or combine the same data of both the dataframes or datasets.

left join: It will combine all the data of the left dataframes and combine the entities which are same in left dataframe and the right dataframe.

right join: complete right dataframe or dataset and matching entities in both left and right dataframe.

outer join: combine both the datasets or dataframes completely.

```python
print("Default Datatypes",df.dtypes)
```

```python
df["sales"] = (df["sales"].astype(str).str.replace(",", "", regex=False).astype(float))
df["order_date"] = pd.to_datetime(df["order_date"],format="mixed",dayfirst=True,errors="coerce")
df["ship_date"] = pd.to_datetime(df["ship_date"],format="mixed",dayfirst=True,errors="coerce")
print(df[["sales", "order_date", "ship_date"]].dtypes)
```
