```python
import pandas as pd
```

```python
df = pd.read_csv("/content/SuperStoreOrders.csv.zip")
df
```

```python
df["sales"] = pd.to_numeric(df["sales"], errors="coerce")
```

**Total sales and profit**

```python
Total_sales = df["sales"].sum()
Total_profit = df["profit"].sum()
print("Total Sales:", Total_sales)
print("Total Profit:", Total_profit)
```

**Which category produces the highest sales and profit**

```python
category_analysis = df.groupby("category")[["sales", "profit"]].sum()
print(category_analysis)
print("Category with Highest Sales:")
print(category_analysis["sales"].idxmax())
print("\nCategory with Highest Profit:")
print(category_analysis["profit"].idxmax())
```

**Which sub-category generates the highest profit**

```python
subcategory_profit = df.groupby("sub_category")["profit"].sum().sort_values(ascending=False)
print(subcategory_profit)
print("\nHighest Profit Sub-Category:")
print(subcategory_profit.idxmax())
```

**Region which generates the highest sales and profit**

```python
region_analysis = df.groupby("region")[["sales", "profit"]].sum()
print(region_analysis)
print("\nRegion with Highest Sales:")
print(region_analysis["sales"].idxmax())
print("\nRegion with Highest Profit:")
print(region_analysis["profit"].idxmax())
```

**Product with highest sales**

```python
product_sales = df.groupby("product_name")["sales"].sum().sort_values(ascending=False)
print(product_sales.head(10))
print("\nProduct with Highest Sales:")
print(product_sales.idxmax())
```

**Year with highest sales and profit**

```python
yearly_analysis = df.groupby("year")[["sales", "profit"]].sum()
print(yearly_analysis)
print("\nYear with Highest Sales:")
print(yearly_analysis["sales"].idxmax())
print("\nYear with Highest Profit:")
print(yearly_analysis["profit"].idxmax())
```

**Category with highest sales in 2014**

```python
sales_2014_category = (df[df["year"] == 2014].groupby("category")["sales"].sum().sort_values(ascending=False))
print(sales_2014_category)
print("Category with highest sales in 2014:",sales_2014_category.idxmax())
```

**sales_2014_region**

```python
sales_2014_region = (df[df["year"] == 2014].groupby("region")["sales"].sum().sort_values(ascending=False))
print(sales_2014_region)
```
