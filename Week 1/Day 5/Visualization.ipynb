**When would you use a bar chart vs a histogram?**

Barchart:rectangular bars that are used to compare numerical values across different categories
Histogram:a graphical data visualization tool that groups numerical data into adjacent intervals (called "bins" or "buckets") to show the frequency distribution of a variable

**What is the difference between plt.plot() and sns.lineplot()?**

plt.plot():it is the basic line plotting
sns.lineplot():Statistical line plotting

**What does a boxplot actually show — what are the whiskers?**

boxplot shows the distribution of numeric data
Median → middle value
Q1 → 25th percentile
Q3 → 75th percentile
Box → middle 50% of data
Whiskers → values within the normal range
Dots beyond whiskers → potential outliers

**How do you plot multiple charts in one figure using Matplotlib?**

by using plt.subplot()

**What is a heatmap useful for and what kind of data does it work best with?**

Heatmap uses color gradients to show the density, high points, and low points of information across a grid or matrix.It usually works with large numerical datasets,categorial data and spatial data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/SuperStoreOrders.csv.zip")
df["sales"] = pd.to_numeric(df["sales"], errors="coerce")
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
```

**Histogram**

```python
plt.figure(figsize=(8,5))
plt.hist(df["sales"].dropna(), bins=30, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.legend(["Sales"])
plt.show()
```

it shows how sales are distributed

**Bar plot**

```python
avg_sales = df.groupby("category")["sales"].mean()
plt.figure(figsize=(8,5))
plt.bar(avg_sales.index, avg_sales.values)

plt.title("Average sale category")
plt.xlabel("category")
plt.ylabel("Average sales")
plt.legend(["Average sales"])
plt.show()
```

this chart compares the average sales of product categories.THe highest bar shows the highest average sale per order

**Box plot**

```python
plt.figure(figsize=(8,5))
sns.boxplot(y=df["sales"].dropna())

plt.title("Sales Boxplot")
plt.xlabel("Sales")
plt.ylabel("Sales_Value")
plt.legend([], [], frameon=False)
plt.show()
```

**Heatmap**

```python
numeric_df = df.select_dtypes(include=np.number)
plt.figure(figsize=(10,7))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap of Numeric Columns")
plt.xlabel("Numeric Variables")
plt.ylabel("Numeric Variables")
plt.show()
```

heatmap shows relation between numeric variables.Values close to +1 indicate a strong positive relationship,Values close to -1 indicate a strong negative relationship,Values close to 0 indicate a weak positive relationship
